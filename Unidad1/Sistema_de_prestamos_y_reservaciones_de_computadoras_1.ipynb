```python
#Sistemas de prestamos
#Programacion basica

#Alumnos que estan en el sistema
alumnos =["26040220", "26040221", "26040222", "26040223", "26040224"]

#Cada caloer en la lista tiene si liste correspondiente con el alunmo
computadoras =["PC-LC1-01", "PC-LC1-02", "PC-LC1-03",
               "PC-LC2-01", "PC-LC2-02", "PC-LC2-03",
               "PC-LC3-01", "PC-LC3-02", "PC-LC3-03", "PC-LC3-04"]
aulas = ["LC1", "LC1", "LC1",
         "LC2", "LC2", "LC2",
         "LC3", "LC3", "LC3", "LC3"]
software =["Python, VS Code", "Python, Java VS Code", "AutoCAD, MATLAB",
           "Python, VS Code", "Python, Java VS Code", "AutoCAD, MATLAB",
           "Python, VS Code", "Python, Java VS Code", "AutoCAD, MATLAB", "Python, VS Code"]
estados = ["Disponible", "Prestada", "Reservada",
           "Disponible", "Prestada", "Reservada",
           "Disponible", "Prestada", "Reservada", "Mantenimiento"]

#MENÚ PRINCIPAL
#USAMNOS EL WHILE PARA QUE SE REPITA INDEFINIDAMENTE HASTA QUE EL USARIA
#TECLEE LA OPCION 7 QUE ES SALIR O CUALQUIER OTRO MAYOR QUE 7
opcion = 0
while opcion <= 6:
  print("=============================")
  print("   SISTEMAS DE PRESTAMOS")
  print("=============================")
  print("1. Consultar computadoras")
  print("2. Buscar por software")
  print("3. Solicitar préstamo")
  print("4. Devolver computadora")
  print("5. Crear reservavión")
  print("6. Agregar computadora")
  print("7. Salir")
  print("=============================")
  opcion = int(input("Seleccione una opción: "))
#1. Consultar computadoras
  if opcion == 1:
    print("-------------------------------------------COMPUTADORAS-----------------------------------------")
    for i in range(len(computadoras)):
      print("          Aula: ", aulas[i], ", ID:", computadoras[i], ", Software:", software[i], ", Estado:", estados[i])
      print("------------------------------------------------------------------------------------------------")
#Buscar se software
  elif opcion == 2:
      programa = input("Ingrese el sofware que quieres buscar: ")
      print("---Resultados---")
      encontrado = False
      for i in range(len(computadoras)):
        if programa.lower() in software[i].lower():
          print("Aula: ", aulas[i], "ID: ", computadoras[i], "Software: ", software[i], "Estado: ", estados[i])
          encontrado = True
      if encontrado == False:
        print("No hay computadoras con ese software")
#Solictar prestamo
  elif opcion == 3:
    print("---Solicitar prestamo---")
    num_control = input("Ingrese su numero de control: ")
    encontrado1 = False
  #Verificar si existe el alumno en la lista BD
    if num_control in alumnos:
    #Si existe, continuamos con el procedimineto
      encontrado1 = True
      id_computadora = input("Ingrese el ID de la computadora: ")
      encontrado2 = False
      for i in range(len(computadoras)):
        if id_computadora == computadoras[i]:
          encontrado2 = True
          if estados[i] == "Disponible":
            estados[i] = "Prestada"
            print("La computadora ha sido prestada")
            print("Computadora:", computadoras[i])
            print("Aula:", aulas[i])
          else:
            print("La computadora no esta disponible")
      if encontrado2 == False:
        print("La computadora no existe")
    else:
      print("El alumno no existe")
    if encontrado2 == False:
      print("La computadora no existe")
      if encontrado1 == False:
        print("El alumno no existe")

  # 6. Agregar computadora
  elif opcion == 6:
    print("\n--- AGREGAR COMPUTADORA ---")
    id_computadora = input("ID de la computadora: ")
    aula = input("Aula: ")
    programas = input("Software instalado: ")

    # En lugar de 4 listas, 1 solo objeto
    nueva_compu = {
        "id": id_computadora,
        "aula": aula,
        "software": programas,
        "estado": "Disponible"
    }
    computadoras.append(nueva_compu)
    print("\nComputadora agregada correctamente.")

  elif opcion==4:
    print("devolver la computadora")
  id_computadora = input("Idame el ID de la computadora: ")
  for i in  range(len(computadoras)):
        if computadoras[i] == id_computadora:
          if estados[i] == "prestada" or estados[i] == "reservada":
            estados[i] = "disponible"
            print("la computadora", id_computadora, "esta disponible")
          else:
            print("la computadora", id_computadora, "no esta prestada")
```
