```python
tabla = int(input("¿Qué tabla de multiplicar quieres?: "))
inicio = int(input("¿Desde qué número deseas empezar?: "))
fin = int(input("¿Hasta qué número deseas terminar?: "))

# Validamos que el fin sea mayor que el inicio antes de avanzar
while inicio >= fin:
    print("El límite final debe de ser mayor que el inicial.")
    fin = int(input("¿Hasta qué número deseas terminar?: "))

# Pedimos los incrementos fuera del while para que siempre existan
incrementos = int(input("¿De cuánto en cuánto?: "))

# Ciclo para mostrar la tabla
for i in range(inicio, fin + 1, incrementos):
    print(tabla, "x", i, "=", tabla * i)
```
